```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark RDD exo").getOrCreate()
```

```python
order_rdd = spark.sparkContext.textFile("./orders_wh.csv")
```

```python
mapped_rdd = order_rdd.map(lambda x: (x.split(",")[3], 1))
```

```python
reduced_rdd = mapped_rdd.reduceByKey(lambda x,y: x + y)
```

```python
reduced_rdd.collect()
```

```python
reduced_rdd.collect()
```

```python
reduced_rdd.cache()
```
