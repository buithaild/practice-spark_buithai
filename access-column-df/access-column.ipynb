```python
from pyspark.sql import SparkSession
spark = SparkSession. \
builder. \
master("local[*]"). \
appName("Truy cap 1 cot trong df"). \
getOrCreate()
```

```python
orders_df = spark.read \
.format("csv") \
.option("header", "true") \
.load("D:\Learn-spark\learn-spark-maide\orders_wh.csv")
```
