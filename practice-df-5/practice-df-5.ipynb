```python
from pyspark.sql import SparkSession
spark = SparkSession. \
builder. \
master("local[*]"). \
appName("Practice df 5: dateformat- pivort"). \
getOrCreate()
```

```python
log_schema = "loglevel string, logtime timestamp"
```

```python
df = spark.read \
.format("csv") \
.schema(log_schema) \
.load("D:\Learn-spark\learn-spark-maide\log_data.csv")
```

```python
df.show()
```

```python
df.createOrReplaceTempView("log_server")
```

```python
spark.sql("select * from log_server limit 10").show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'M') as month from log_server").show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'MM') as month from log_server").show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'MMM') as month from log_server").show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'MMMM') as month from log_server").show()
```

```python
#Số lần xuất hiện log trong tháng
spark.sql("select loglevel, date_format(logtime, 'MMMM') as month, count(*) as total from log_server group by loglevel, month order by month").show()
```

```python
spark.sql("""select loglevel, 
          date_format(logtime, 'MMMM') as month,
          date_format(logtime, 'M') as month_num, 
          count(*) as total 
          from log_server 
          group by loglevel, month, month_num
          order by month_num""").show()
# month_num lúc này hiểu là text
```

```python
spark.sql("""select loglevel, 
          date_format(logtime, 'MMMM') as month,
          int(date_format(logtime, 'M')) as month_num, 
          count(*) as total 
          from log_server 
          group by loglevel, month, month_num
          order by month_num""").show()
```

```python
spark.sql("""select loglevel, 
          date_format(logtime, 'MMMM') as month,
          date_format(logtime, 'MM') as month_num, 
          count(*) as total 
          from log_server 
          group by loglevel, month, month_num
          order by month_num""").show()
```

```python
df = spark.sql("""select loglevel, 
          date_format(logtime, 'MMMM') as month,
          date_format(logtime, 'MM') as month_num, 
          count(*) as total 
          from log_server 
          group by loglevel, month, month_num
          order by month_num""")
```

```python
df.show()
```

```python
df_final = df.drop("month_num")
```

PIVOT

```python
spark.sql("select loglevel, date_format(logtime, 'MMMM') as month from log_server").show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'MMMM') as month from log_server") \
    .groupBy("loglevel").pivot('month').count() \
    .show()
```

```python
spark.sql("select loglevel, date_format(logtime, 'MM') as month from log_server") \
    .groupBy("loglevel").pivot('month').count() \
    .show()
```

```python
month_list = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
```

```python
spark.sql("select loglevel, date_format(logtime, 'MMMM') as month from log_server") \
    .groupBy("loglevel").pivot('month', month_list).count() \
    .show()
```
