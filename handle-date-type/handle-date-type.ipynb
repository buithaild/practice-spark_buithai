```python
from pyspark.sql import SparkSession

spark = SparkSession. \
builder. \
appName("Hanle Date Type"). \
getOrCreate()
```

```python
order_schema = "order_id long, order_date date, cust_id long, status string"
```

YYYY-MM-DD

```python
df = spark.read \
.format("csv") \
.schema(order_schema) \
.load("D:\Learn-spark\learn-spark-maide\order_sample_10line.csv")
```

```python
df.show()
```

```python
df.printSchema()
```
