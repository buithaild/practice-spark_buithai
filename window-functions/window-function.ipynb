```python
from pyspark.sql import SparkSession
spark = SparkSession. \
builder. \
master("local[*]"). \
appName("Window functions"). \
getOrCreate()
```

```python
df = spark.read \
.format("csv") \
.option ("inferSchema", "true") \
.option("header", "true") \
.load("D:\Learn-spark\learn-spark-maide\weekly_invoices_small.csv")
```

```python
df.show()
```

```python
df.sort("Country").show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
```

```python
my_window = Window.partitionBy("Country").orderBy("InvoiceValue")
```

```python
#rank
df.withColumn("rank", rank().over(my_window)).show(100)
```

```python
df.withColumn("dense_rank", dense_rank().over(my_window)).show(100)
```

```python
df.withColumn("row_number", row_number().over(my_window)).show(100)
```

```python
df1 = df.withColumn("row_number", row_number().over(my_window))
```

```python
#Tuần mà danh số bán hàng thấp nhất cua tung quoc gia
df1.where("row_number = 1").show()
```

```python
#Xem doanh so cao nhat thi sap xep giam dan
my_window_2 = Window.partitionBy("Country").orderBy(desc("InvoiceValue"))
```

```python
df.withColumn("row_number", row_number().over(my_window_2)).show()
```

```python
df.withColumn("row_number", row_number().over(my_window_2)).where("row_number = 1").show()
# df.withColumn("row_number", row_number().over(my_window_2)).where("row_number = 1").drop("row_number").show()
```

LEAD; LAG

```python
df.sort("Country").show()
```

```python
my_win = Window.partitionBy("Country").orderBy("WeekNum")
```

```python
df.withColumn("previous_week", lag("InvoiceValue").over(my_win)) \
    .withColumn("diff_invoice", expr("InvoiceValue - previous_week")) \
    .show()
```

```python
df.withColumn("next_week_invoice", lead("InvoiceValue").over(my_win)) \
    .withColumn("diff_invoice", expr("InvoiceValue - next_week_invoice")) \
    .show()
```
