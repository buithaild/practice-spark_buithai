```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("Spark RDD exo").getOrCreate()
```

```python
#Buoc 1: Load du lieu
order_rdd = spark.sparkContext.textFile("./orders_wh.csv")
```

```python
#Buoc 2- Ham map: Narrow transformations
mapped_rdd = order_rdd.map(lambda x: (x.split(",")[3], 1))
```

```python
#Buoc 3- Ham reduceByKey: Wipe transformations
reduced_rdd = mapped_rdd.reduceByKey(lambda x,y: x+y)
```

```python
#Buoc 4- collect: Actions
reduced_rdd.collect()
```

```python
print(reduced_rdd.getNumPartitions())
```
